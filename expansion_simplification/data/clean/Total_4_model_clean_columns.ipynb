# Expansion and Simplification Clean Dataset Column Definitions

This notebook describes the columns in `Total_4_model_clean.jsonl`. Each JSONL row contains one model response to one polynomial expansion and simplification equation.

| Column | Description |
|---|---|
| `id` | The unique identifier assigned to the Batch API response. It identifies the returned response record rather than the source equation. |
| `model` | The identifier of the model that generated the response. This field distinguishes records from DeepSeek, Gemma, OpenAI, and Qwen in the combined dataset. |
| `question_id` | The source request identifier, such as `question-000001`. The same identifier links corresponding equations across the four model outputs. |
| `question` | The polynomial expansion and simplification prompt returned in the model's structured response. It is `null` when a usable question could not be extracted. |
| `steps` | The model's explanation or calculation steps, normalized as a list of strings. It is `null` when valid steps could not be extracted. |
| `final_answer` | The final expression reported by the model, or the value extracted from `\boxed{...}` in the last step when the explicit answer is missing. It is `null` when no usable answer was available. |
| `whether_correct` | The correctness result obtained by comparing `final_answer` with the exact reference expression. `true` means correct, `false` means wrong, and `null` means the response could not be scored because an answer was missing or unusable. |
