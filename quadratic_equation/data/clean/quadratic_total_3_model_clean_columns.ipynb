# Quadratic Clean Dataset Column Definitions

This notebook describes the columns in `quadratic_total_3_model_clean.jsonl`. Each JSONL row contains one model response to one quadratic equation.

| Column | Description |
|---|---|
| `id` | The unique identifier assigned to the Batch API response. It identifies the returned response record rather than the source equation. |
| `model` | The identifier of the model that generated the response. This field distinguishes records from DeepSeek, OpenAI, and Qwen in the combined dataset. |
| `question_id` | The source request identifier, such as `synthetic_0`. The same identifier links corresponding equations across the three model outputs. |
| `question` | The quadratic-equation prompt returned in the model's structured response. It is `null` when the response payload could not be parsed. |
| `steps` | The model's explanation or calculation steps, normalized as a list of strings. It is `null` when valid steps could not be extracted. |
| `final_answer` | The real root or roots reported by the model, or `No real solution` when the model concluded that no real roots exist. It is `null` when no usable answer was available. |
| `whether_correct` | The correctness result obtained by comparing `final_answer` with the exact reference solutions. `true` means correct, `false` means wrong, and `null` means the response could not be scored because an answer was missing or unusable. |
